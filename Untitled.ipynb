## 1. Import Libraries

```python
import numpy as np
import pandas as pd
import tensorflow as tf
```

## 2. Load Dataset

```python
dataset = pd.read_csv("bodyfat.csv")
dataset.head()
```

## 3. Split Features & Target

```python
X = dataset.drop('BodyFat', axis=1).values
y = dataset['BodyFat'].values
```

```python
print(X)
```

```python
print(y)
```

```python
dataset.columns
```

```python
dataset.info()
```

```python
# No Encoding Needed (all columns are numerical)
```

## 4. Train Test Split

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)
```

## 5. Feature Scaling

```python
from sklearn.preprocessing import StandardScaler
```

```python
sc = StandardScaler()
```

```python
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

## 6. Build ANN

```python
ann = tf.keras.models.Sequential()
```

```python
# Adding input layer and hidden layers
ann.add(tf.keras.layers.Dense(64, activation='relu'))
ann.add(tf.keras.layers.Dense(128, activation='relu'))
ann.add(tf.keras.layers.Dense(64, activation='relu'))
ann.add(tf.keras.layers.Dense(32, activation='relu'))
```

```python
# Adding output layer
ann.add(tf.keras.layers.Dense(1))
```

## 7. Compile

```python
ann.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),loss='mean_squared_error',metrics=['mae'])
```

```python
'''
| Case                   | Loss                | Output Layer  | Metric   |
| ---------------------- | ------------------- | ------------- | -------- |
| Classification         | binary_crossentropy | sigmoid       | accuracy |
| Regression (your case) | mean_squared_error  | no activation | R² / MAE |
'''
```

## 8. Train Model

```python
history = ann.fit(X_train, y_train, epochs=100, batch_size=32, validation_split=0.2)
```

## 9. Plot Graphs

```python
import matplotlib.pyplot as plt
plt.plot(history.history['loss'], label='Train Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title('Model Loss')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()
plt.show()
```

## 10. Evaluate Model

```python
y_pred = ann.predict(X_test)
```

```python
y_pred
```

```python
from sklearn.metrics import r2_score, mean_absolute_error
```

```python
print("R2 Score:", r2_score(y_test, y_pred))
print("MAE:", mean_absolute_error(y_test, y_pred))
```
